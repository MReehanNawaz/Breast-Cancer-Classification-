```python
# Single layer perceptron Breast Cancer Classification

import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
import tensorflow as tf 
from sklearn.datasets import load_breast_cancer 


cancer_dataset = load_breast_cancer()
x = cancer_dataset.data
y = cancer_dataset.target
Feature = cancer_dataset.feature_names
cancer_df = pd.DataFrame(x, columns=Feature )
print(Feature)

scaler = StandardScaler()
x = scaler.fit_transform(x)
cancer_df.head()
```

```python
print('Shape : ', cancer_df.shape)
print('X_shape : ', x.shape)

print('Y_shape : ', y.shape, type(y))
print(y)

y = y.reshape(-1, 1)
onehotencoder = OneHotEncoder()
y = onehotencoder.fit_transform(y).toarray()
print('\ny Seperate cols')
print('y',y)


# split the dataset 

x_train , x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=21
)



# Create tensorflow data 

x_train_tf = tf.constant(x_train, dtype=tf.float32)
y_train_tf = tf.constant(y_train, dtype=tf.float32)

# 7. Model parameters
learning_rate = 0.01 
n_features = 30
n_class = 2 

# Initialize weights and bias 

weight = tf.Variable(tf.zeros([n_features, n_class]), dtype=tf.float32)
Bias = tf.Variable(tf.zeros([n_class]), dtype=tf.float32)


# 9. Optimizer 

optimizer = tf.keras.optimizers.SGD(learning_rate = learning_rate)

for epoch in range(100):
    with tf.GradientTape() as tape :
        # Calculation 

        z = tf.matmul(x_train_tf, weight) + Bias 

        # Convert into probability 

        probb = tf.nn.softmax(z)

        # Calculate Loss 

        loss = tf.reduce_mean(
            tf.keras.losses.categorical_crossentropy(y_train_tf, probb)
        )

     # Calculate gradients 
    dw, db = tape.gradient(loss, [weight, Bias])


    # Update the weights and Bias 

    optimizer.apply_gradients([(dw, weight), (db, Bias)])

    if (epoch + 1) % 10 == 0:
            print(
                "Epoch:",
                epoch + 1,
                "Loss:",
                loss.numpy()
            )
        
# Training loss
# Test loss
# Test accuracy
# Confusion matrix

# Convert test data to TensorFlow tensor

x_test_tf = tf.constant(x_test, dtype=tf.float32)

# Make predictions
z_test = tf.matmul(x_test_tf, weight) + Bias
probabilities = tf.nn.softmax(z_test)

# Get predicted class
y_pred = tf.argmax(probabilities, axis=1)

# Get actual class
y_actual = tf.argmax(y_test, axis=1)

print("Predicted:", y_pred.numpy())
print("Actual   :", y_actual.numpy())

accuracy = tf.reduce_mean(
    tf.cast(y_pred == y_actual, tf.float32)
)

print("Test Accuracy:", round(accuracy.numpy(),2))
```

```python
from sklearn.metrics import confusion_matrix, classification_report
cm = confusion_matrix(
    y_actual.numpy(),
    y_pred.numpy()
)

print("Confusion Matrix:")
print(cm)

print(classification_report(
    y_actual.numpy(),
    y_pred.numpy()
))
```
